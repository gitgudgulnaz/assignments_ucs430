## Q6. Create the employees dataset

```python
emp_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}
emp = pd.DataFrame(emp_data)
emp.to_csv("employees.csv", index=False)
emp
```

### e) Statistics

```python
avg_salary = emp["Salary"].mean()
total_bonus = emp["Bonus"].sum()
youngest_age = emp["Age"].min()
highest_rating = emp["Rating"].max()

print("i.   Average salary:", avg_salary)
print("ii.  Total bonus paid:", total_bonus)
print("iii. Youngest employee's age:", youngest_age)
print("iv.  Highest performance rating:", highest_rating)
```

### f) Sort the DataFrame by Salary, descending

```python
emp_sorted = emp.sort_values(by="Salary", ascending=False)
emp_sorted
```

### g) Add a Performance_Category column

```python
def categorize(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

emp["Performance_Category"] = emp["Rating"].apply(categorize)
emp
```

### h) Identify missing values

```python
emp.isnull().sum()
```

### i) Rename Employee_ID to ID

```python
emp = emp.rename(columns={"Employee_ID": "ID"})
emp.columns.tolist()
```

### j) Filter employees

```python
# i. More than 5 years of experience
exp_gt_5 = emp[emp["Years_of_Experience"] > 5]
exp_gt_5
```

```python
# ii. Belong to the IT department
it_dept = emp[emp["Department"] == "IT"]
it_dept
```

### k) Add a Tax column — 10% of Salary

```python
emp["Tax"] = emp["Salary"] * 0.10
emp
```

### l) Save the modified DataFrame to a new CSV file

```python
emp.to_csv("employees_modified.csv", index=False)
print("Saved to employees_modified.csv")
```

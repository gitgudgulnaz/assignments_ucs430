```python
n = int(input("enter your number: "))
prime_sum = 0
for num in range(2, n + 1):
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        prime_sum = prime_sum+num
print(f"the sum of all prime numbers from 1 to {n} is: {prime_sum}")
#1 used f string in last line
#2 1 is not a prime number
#3 also didnt have to check all numbers in the loop, just need to check until the root of the number (for optimisation)
#4 for some reason in the github example of prime number function(6.2) they used n//2 + 1
```

```python
n = int(input("enter your number: "))
def odd(n):
    odd_sum = 0
    for i in range(1, n+1, 2):
        odd_sum += i
    return odd_sum
print(f"{odd(n)}")
```

```python
def isprime(num):
    if num < 2:
        return False
    for i in range(2 , int(num ** 0.5) + 1):
        if num%i == 0:
            return False
    return True
def prime_s(n):
    prime_sum = 0
    for i in range(1, n+1):
        if isprime(i):
            prime_sum += i
    return prime_sum
n = int(input("enter your no."))
print(prime_s(n))
```

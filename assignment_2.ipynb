```python
roll_number = "1024170109"
L = [int(digit) * 10 for digit in roll_number]
print("i. initial list L:", L)
```

```python
L.append(50)  # append() adds 50 to the end of the list
print("ii. after append(50):", L)

L.insert(2, 25)  # insert() places 25 at index position 2
print(" after insert(2, 25):", L)
```

```python
L.remove(0)  # remove() deletes the first occurrence of value 0
print("iii. after remove(0):", L)

popped_val = L.pop(4)  # pop(4) removes and returns the element at index 4
print(f"after pop(4) [removed {popped_val}]:", L)
```

```python
L.sort()
print("iv. ascending order:", L)

L.sort(reverse=True)
print("descending order:", L)
```

```python
print("v. first three elements:", L[:3])
print("last three elements:", L[-3:])
```

```python
avg = sum(L) / len(L)
above_avg = [x for x in L if x > avg]
print(f"vi. average of L: {avg:.2f}")
print("elements greater than average:", above_avg)
```

```python
# Using the first 8 values from L: [10, 0, 20, 40, 10, 70, 0, 10]
scores = (10, 0, 20, 40, 10, 70, 0, 10)

max_score = max(scores)
max_index = scores.index(max_score)
min_score = min(scores)
min_count = scores.count(min_score)

print("i. highest score:", max_score, "at index:", max_index)
print("lowest score:", min_score, "appearing", min_count, "times")

reversed_scores_list = list(reversed(scores)) # Tuples are immutable so their items cannot be reordered in place like lists.
print("ii. Reversed tuple as list:", reversed_scores_list)

user_score = int(input("iii. Enter a score to search: "))
if user_score in scores:
    print(f"Score {user_score} first appears at index {scores.index(user_score)}")
else:
    print(f"Score {user_score} is not present in the tuple.")

try:
    scores[0] = 100
except TypeError as e:
    print("iv. error captured:", e) # Tuples are immutable and lack item assignment, unlike mutable lists which allow direct updates.

first, second, *remaining = scores
print("v. first score:", first)
print("   second score:", second)
print("   remaining scores:", remaining)
```

```python
import random
from collections import Counter

roll_number = 1024170109
random.seed(roll_number)

numbers = [random.randint(100, 900) for i in range(100)]

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

odds = [x for x in numbers if x % 2 != 0]
print("ii. count of odd numbers:", len(odds))

evens = [x for x in numbers if x % 2 == 0]
print("iii. count of even numbers:", len(evens))

primes = [x for x in numbers if is_prime(x)]
print("iv. count of prime numbers:", len(primes))
print("prime numbers found:", primes)

counts = Counter(numbers)
most_common_num, most_common_count = counts.most_common(1)[0]
print(f"v. most frequent number: {most_common_num} (occurs {most_common_count} times)")
```

```python
digits = [1, 0, 2, 4, 1, 7, 0, 1]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("set A:", A)
print("set B:", B)

union_set = A.union(B)
print("vi. union of A and B:", union_set)

intersection_set = A.intersection(B)
print("vii. intersection of A and B:", intersection_set)

diff_a_b = A.difference(B)
diff_b_a = B.difference(A)
print("viii. A - B:", diff_a_b)
print("B - A:", diff_b_a) # difference() returns elements in one set but not the other, whereas symmetric_difference() returns all unique elements from both sets excluding their intersection.

sym_diff = A.symmetric_difference(B)
print("ix. symmetric difference of A and B:", sym_diff)

is_a_sub_b = A.issubset(B)
is_b_super_a = B.issuperset(A)
print("x. is A a subset of B?:", is_a_sub_b)
print("is B a superset of A?:", is_b_super_a)

X = int(input("xi. Enter a value X to discard from A: "))
A.discard(X) # discard() removes an element if present without raising a KeyError if it is missing, unlike remove() which crashes.
print(f"set A after discarding {X}:", A)
```
